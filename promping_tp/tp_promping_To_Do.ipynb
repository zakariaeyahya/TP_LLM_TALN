```python
from openai import OpenAI


client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key="insert you key",  
)

def send_prompt(prompt, model="deepseek/deepseek-r1-distill-llama-70b:free", temperature=0.7):
    """
    Envoie un prompt au modèle et renvoie la réponse textuelle.
    """
    try:
        response = client.chat.completions.create(
            model=model,
            messages=[{"role": "user", "content": prompt}],
            temperature=temperature,
        )

        # DEBUG pour voir la structure
        print("DEBUG Response:", response)

        # Accès correct à la réponse
        return response.choices[0].message.content

    except Exception as e:
        print(f"Erreur lors de la génération : {e}")
        return None
```

**zero shot**

Zero-shot Prompting :Est un type de prompt où tu donnes une instruction sans aucun exemple.
Le modèle doit deviner tout seul la meilleure manière de répondre.

```python
def main(prompt):
    print("=== PROMPT ENVOYÉ ===")
    print(prompt)
    print("\n=== RÉPONSE DU MODÈLE ===\n")

    result = send_prompt(prompt)
    if result:
        print(result)
    else:
        print("Aucune réponse reçue.")


if __name__ == "__main__":
    #TO DO: Écrire un prompt ZERO-SHOT qui demande au modèle
    # d'écrire un post Instagram inspirant sur la productivité.
    PROMPT = ""
    main(PROMPT)
```
